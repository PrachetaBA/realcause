```python
import pandas as pd
import sqlalchemy as sa

engine = sa.create_engine(database)

with engine.connect() as conn:
    table_names = pd.read_sql(
        "SELECT name FROM sqlite_master WHERE type='table' ORDER BY name",
        conn,
    )
    sample_schema = pd.read_sql("PRAGMA table_info(sample)", conn)

table_names, sample_schema
```

```python
# Write code to now read the populations table from the database and convert it to a pandas DataFrame
populations_df = pd.read_sql("SELECT * FROM populations", engine)
print(populations_df)
```

```python
# Read the sample table from the database and convert it to a pandas DataFrame
sample_df = pd.read_sql("SELECT * FROM samples", engine)
print(sample_df)
```

```python
# read the models table from the database and convert it to a pandas DataFrame
models_df = pd.read_sql("SELECT * FROM models", engine)
print(models_df)
```
